```python
# ERAG and LLM Based Severity Prediction of Bug reports Code 
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from transformers import GPT2Tokenizer, GPT2ForSequenceClassification, GPT2Config, GPT2LMHeadModel
from transformers import get_linear_schedule_with_warmup
from torch.optim import AdamW
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, roc_curve, auc
from sklearn.preprocessing import label_binarize
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import matplotlib.pyplot as plt
import seaborn as sns
import os
import random
import re
from collections import defaultdict

# Set device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# Hyperparameters
MAX_LENGTH = 256
BATCH_SIZE = 16
EPOCHS = 15  # Increased epochs for better learning
LEARNING_RATE = 2e-5
WARMUP_STEPS = 100
SAMPLES_PER_CLASS = 250
DROPOUT_RATE = 0.2  # Slightly increased dropout for better regularization
RAG_TOP_K = 3  # Number of similar examples to retrieve
CLASS_NAMES = ['Critical', 'High', 'Medium', 'Low', 'Trivial']  # More descriptive names

# Set environment variable to avoid tokenizer parallelism warning
os.environ["TOKENIZERS_PARALLELISM"] = "false"

# RAG Knowledge Base class
class RAGKnowledgeBase:
    def __init__(self):
        self.examples_by_class = defaultdict(list)
        self.vectorizer = TfidfVectorizer(max_features=5000, stop_words='english')
        self.vectors = None
        self.texts = []
        self.labels = []
    
    def add_examples(self, texts, labels):
        """Add examples to the knowledge base"""
        for text, label in zip(texts, labels):
            self.examples_by_class[label].append(text)
            self.texts.append(text)
            self.labels.append(label)
        
        # Update TF-IDF vectors
        if self.texts:
            self.vectors = self.vectorizer.fit_transform(self.texts)
    
    def retrieve_similar(self, query_text, top_k=3):
        """Retrieve similar examples from the knowledge base"""
        if not self.texts or self.vectors is None:
            return []
        
        # Vectorize query
        query_vec = self.vectorizer.transform([query_text])
        
        # Calculate similarities
        similarities = cosine_similarity(query_vec, self.vectors).flatten()
        
        # Get top k most similar examples
        top_indices = similarities.argsort()[-top_k:][::-1]
        
        retrieved_examples = []
        for idx in top_indices:
            retrieved_examples.append({
                'text': self.texts[idx],
                'label': self.labels[idx],
                'similarity': similarities[idx]
            })
        
        return retrieved_examples

# Load and preprocess data
def preprocess_data(df):
    # Handle missing values
    df = df.dropna(subset=['preprocessed', 'review'])
    
    # Use preprocessed text if available, otherwise use original review
    df['text'] = df['preprocessed'].fillna(df['review'])
    
    # Convert stars to 0-4 scale (assuming stars are 1-5, subtract 1)
    # But since you mentioned they're already 0-4, we'll keep as is
    if 'star' in df.columns:
        df['label'] = df['star']
    elif 'label' in df.columns:
        df['label'] = df['label']
    else:
        raise ValueError("DataFrame must contain either 'star' or 'label' column")
    
    # Filter out invalid labels (0-4)
    df = df[df['label'].between(0, 4)]
    
    return df

# Custom Dataset class with RAG support
class BugReportDataset(Dataset):
    def __init__(self, texts, labels, tokenizer, max_length, rag_kb=None, use_rag=False):
        self.texts = texts
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_length = max_length
        self.rag_kb = rag_kb
        self.use_rag = use_rag
    
    def __len__(self):
        return len(self.texts)
    
    def __getitem__(self, idx):
        text = str(self.texts[idx])
        label = self.labels[idx]
        
        # If using RAG, retrieve similar examples and format the input
        if self.use_rag and self.rag_kb:
            retrieved = self.rag_kb.retrieve_similar(text, top_k=RAG_TOP_K)
            
            # Format the input with retrieved examples
            prompt = "Classify the severity of this bug report based on similar examples:\n\n"
            
            for i, example in enumerate(retrieved):
                severity = CLASS_NAMES[example['label']]
                prompt += f"Example {i+1} ({severity}): {example['text']}\n\n"
            
            prompt += f"Bug report to classify: {text}\n\nSeverity:"
            text = prompt
        
        encoding = self.tokenizer(
            text,
            truncation=True,
            padding='max_length',
            max_length=self.max_length,
            return_tensors='pt'
        )
        
        return {
            'input_ids': encoding['input_ids'].flatten(),
            'attention_mask': encoding['attention_mask'].flatten(),
            'labels': torch.tensor(label, dtype=torch.long)
        }

# Initialize model with proper padding token handling
def initialize_model(num_labels=5, tokenizer=None, use_rag=False):
    # Load pre-trained GPT-2
    if use_rag:
        # For RAG approach, we'll use the LM head and do classification via generation
        model = GPT2LMHeadModel.from_pretrained('gpt2')
    else:
        # For standard classification
        config = GPT2Config.from_pretrained(
            'gpt2',
            num_labels=num_labels,
            problem_type="single_label_classification",
            pad_token_id=tokenizer.pad_token_id if tokenizer else 0
        )
        
        model = GPT2ForSequenceClassification.from_pretrained(
            'gpt2',
            config=config
        )
    
    # Set padding token in model config
    if tokenizer:
        model.config.pad_token_id = tokenizer.pad_token_id
        # Resize token embeddings to include padding token
        model.resize_token_embeddings(len(tokenizer))
    
    # Add dropout for regularization
    model.config.hidden_dropout_prob = DROPOUT_RATE
    model.config.attention_probs_dropout_prob = DROPOUT_RATE
    
    return model.to(device)

# Training function with class weights to handle imbalance
def train_model(model, train_loader, val_loader, optimizer, scheduler, epochs, use_rag=False):
    model.train()
    best_accuracy = 0
    train_losses = []
    val_accuracies = []
    
    # Add class weights to handle potential imbalance
    class_weights = torch.tensor([1.0, 1.0, 1.0, 1.5, 2.0]).to(device)  # Higher weights for Low and Cosmetic classes
    criterion = nn.CrossEntropyLoss(weight=class_weights) if not use_rag else None
    
    for epoch in range(epochs):
        total_loss = 0
        model.train()
        
        for batch_idx, batch in enumerate(train_loader):
            optimizer.zero_grad()
            
            input_ids = batch['input_ids'].to(device)
            attention_mask = batch['attention_mask'].to(device)
            labels = batch['labels'].to(device)
            
            if use_rag:
                # For RAG approach, we need to generate text and then extract the classification
                outputs = model(
                    input_ids=input_ids,
                    attention_mask=attention_mask,
                    labels=input_ids  # For language modeling, we use input as labels
                )
                loss = outputs.loss
            else:
                # Standard classification with weighted loss
                outputs = model(
                    input_ids=input_ids,
                    attention_mask=attention_mask
                )
                loss = criterion(outputs.logits, labels)
            
            loss.backward()
            
            # Gradient clipping
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            
            optimizer.step()
            scheduler.step()
            
            total_loss += loss.item()
            
            if batch_idx % 50 == 0:
                print(f'Epoch {epoch+1}/{epochs} | Batch {batch_idx}/{len(train_loader)} | Loss: {loss.item():.4f}')
        
        avg_loss = total_loss / len(train_loader)
        train_losses.append(avg_loss)
        
        # Validation
        val_accuracy = evaluate_model(model, val_loader, use_rag)
        val_accuracies.append(val_accuracy)
        
        print(f'Epoch {epoch+1}/{epochs} | Avg Loss: {avg_loss:.4f} | Val Accuracy: {val_accuracy:.4f}')
        
        # Save best model
        if val_accuracy > best_accuracy:
            best_accuracy = val_accuracy
            torch.save(model.state_dict(), 'best_gpt2_model.pth')
    
    return train_losses, val_accuracies

# Evaluation function
def evaluate_model(model, data_loader, use_rag=False):
    model.eval()
    predictions = []
    actual_labels = []
    
    with torch.no_grad():
        for batch in data_loader:
            input_ids = batch['input_ids'].to(device)
            attention_mask = batch['attention_mask'].to(device)
            labels = batch['labels'].to(device)
            
            if use_rag:
                # For RAG approach, generate text and extract classification
                outputs = model.generate(
                    input_ids=input_ids,
                    attention_mask=attention_mask,
                    max_length=MAX_LENGTH + 50,  # Allow for longer generation
                    num_return_sequences=1,
                    temperature=0.7,
                    do_sample=False,
                    pad_token_id=model.config.pad_token_id
                )
                
                # Extract predictions from generated text
                batch_preds = []
                for i in range(outputs.shape[0]):
                    generated_text = tokenizer.decode(outputs[i], skip_special_tokens=True)
                    # Extract the severity from the generated text
                    pred = extract_severity_from_text(generated_text)
                    batch_preds.append(pred)
                
                predictions.extend(batch_preds)
            else:
                # Standard classification
                outputs = model(
                    input_ids=input_ids,
                    attention_mask=attention_mask
                )
                
                _, preds = torch.max(outputs.logits, dim=1)
                predictions.extend(preds.cpu().tolist())
            
            actual_labels.extend(labels.cpu().tolist())
    
    accuracy = accuracy_score(actual_labels, predictions)
    return accuracy

# Improved function to extract severity from generated text
def extract_severity_from_text(text):
    # Look for severity keywords in the generated text
    text_lower = text.lower()
    
    # Check for cosmetic terms first (most specific)
    cosmetic_terms = ['cosmetic', 'typo', 'spelling', 'grammar', 'formatting', 'ui', 'appearance', 'visual']
    if any(term in text_lower for term in cosmetic_terms):
        return 4
    
    # Check for low severity terms
    low_terms = ['low', 'minor', 'trivial', 'insignificant', 'small', 'tiny']
    if any(term in text_lower for term in low_terms):
        return 3
    
    # Check for medium severity terms
    medium_terms = ['medium', 'moderate', 'average', 'tolerable', 'acceptable']
    if any(term in text_lower for term in medium_terms):
        return 2
    
    # Check for high severity terms
    high_terms = ['high', 'major', 'serious', 'significant', 'important', 'critical']
    if any(term in text_lower for term in high_terms):
        return 1
    
    # Check for critical severity terms
    critical_terms = ['critical', 'crash', 'fatal', 'blocker', 'blocking', 'emergency', 'urgent']
    if any(term in text_lower for term in critical_terms):
        return 0
    
    # If no severity found, return the most common class (Medium)
    return 2

# Get predictions and probabilities for detailed analysis
def get_predictions_and_probabilities(model, data_loader, use_rag=False):
    model.eval()
    predictions = []
    probabilities = []
    actual_labels = []
    
    with torch.no_grad():
        for batch in data_loader:
            input_ids = batch['input_ids'].to(device)
            attention_mask = batch['attention_mask'].to(device)
            labels = batch['labels'].to(device)
            
            if use_rag:
                # For RAG approach, we can't easily get probabilities
                # So we'll use a dummy probability distribution
                outputs = model.generate(
                    input_ids=input_ids,
                    attention_mask=attention_mask,
                    max_length=MAX_LENGTH + 50,
                    num_return_sequences=1,
                    temperature=0.7,
                    do_sample=False,
                    pad_token_id=model.config.pad_token_id
                )
                
                batch_preds = []
                batch_probs = []
                for i in range(outputs.shape[0]):
                    generated_text = tokenizer.decode(outputs[i], skip_special_tokens=True)
                    pred = extract_severity_from_text(generated_text)
                    batch_preds.append(pred)
                    
                    # Create a dummy probability distribution
                    prob = np.zeros(len(CLASS_NAMES))
                    prob[pred] = 1.0
                    batch_probs.append(prob)
                
                predictions.extend(batch_preds)
                probabilities.extend(batch_probs)
            else:
                # Standard classification
                outputs = model(
                    input_ids=input_ids,
                    attention_mask=attention_mask
                )
                
                # Apply softmax to get probabilities
                probs = torch.nn.functional.softmax(outputs.logits, dim=1)
                _, preds = torch.max(outputs.logits, dim=1)
                
                predictions.extend(preds.cpu().tolist())
                probabilities.extend(probs.cpu().tolist())
            
            actual_labels.extend(labels.cpu().tolist())
    
    return np.array(predictions), np.array(probabilities), np.array(actual_labels)

# Create balanced sample data without noise
def create_balanced_sample_data(samples_per_class=250):
    texts = []
    labels = []
    
    # Class 0 samples (Critical) - more specific terms
    class_0_samples = [
        f"Application crashes on startup with fatal error {i}. All users affected. No workaround available. System completely unusable."
        for i in range(samples_per_class)
    ]
    texts.extend(class_0_samples)
    labels.extend([0] * samples_per_class)
    
    # Class 1 samples (High) - major functionality issues
    class_1_samples = [
        f"Major functionality broken in module {i}. Significant impact on user experience. Workaround exists but is cumbersome. Core feature affected."
        for i in range(samples_per_class)
    ]
    texts.extend(class_1_samples)
    labels.extend([1] * samples_per_class)
    
    # Class 2 samples (Medium) - moderate impact
    class_2_samples = [
        f"Feature {i} not working as expected. Moderate impact on some users. Alternative approaches available. Not a critical issue."
        for i in range(samples_per_class)
    ]
    texts.extend(class_2_samples)
    labels.extend([2] * samples_per_class)
    
    # Class 3 samples (Low) - minor issues
    class_3_samples = [
        f"Minor UI issue in screen {i}. Small visual problem only. Does not affect functionality. Low priority."
        for i in range(samples_per_class)
    ]
    texts.extend(class_3_samples)
    labels.extend([3] * samples_per_class)
    
    # Class 4 samples (Cosmetic) - very specific cosmetic issues
    class_4_samples = [
        f"Typo in error message {i}. Spelling mistake in dialog box. Very minor issue. Does not affect any functionality. Purely cosmetic."
        for i in range(samples_per_class)
    ]
    texts.extend(class_4_samples)
    labels.extend([4] * samples_per_class)
    
    # Introduce 1% noise by randomly mislabeling some samples
    total_samples = len(labels)
    num_noisy_samples = int(total_samples * 0.01)  # 1% noise
    
    # Get random indices to add noise
    noisy_indices = random.sample(range(total_samples), num_noisy_samples)
    
    for idx in noisy_indices:
        original_label = labels[idx]
        # Generate a random different label (0-4)
        possible_labels = [l for l in range(5) if l != original_label]
        new_label = random.choice(possible_labels)
        labels[idx] = new_label
    
    # Create DataFrame
    df = pd.DataFrame({
        'preprocessed': texts,
        'review': texts,
        'label': labels
    })
    
    return df

# Plot confusion matrix with percentages
def plot_confusion_matrix_percentage(y_true, y_pred, classes, title='Confusion Matrix'):
    cm = confusion_matrix(y_true, y_pred)
    cm_percentage = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis] * 100
    
    plt.figure(figsize=(10, 8))
    sns.heatmap(cm_percentage, annot=True, fmt='.2f', cmap='Blues', 
                xticklabels=classes, yticklabels=classes)
    plt.title(title)
    plt.ylabel('True Label')
    plt.xlabel('Predicted Label')
    plt.tight_layout()
    plt.savefig('confusion_matrix_percentage.png')
    plt.show()
    
    return cm

# Plot ROC curves for each class with smooth curves
def plot_roc_curves(y_true, y_prob, n_classes, class_names):
    # Binarize the output
    y_true_bin = label_binarize(y_true, classes=range(n_classes))
    
    # Compute ROC curve and ROC area for each class
    fpr = dict()
    tpr = dict()
    roc_auc = dict()
    
    for i in range(n_classes):
        fpr[i], tpr[i], _ = roc_curve(y_true_bin[:, i], y_prob[:, i])
        
        # Interpolate to create smooth curves
        fpr_interp = np.linspace(0, 1, 100)
        tpr_interp = np.interp(fpr_interp, fpr[i], tpr[i])
        tpr_interp[0] = 0.0
        
        fpr[i] = fpr_interp
        tpr[i] = tpr_interp
        roc_auc[i] = auc(fpr[i], tpr[i])
    
    # Plot all ROC curves
    plt.figure(figsize=(10, 8))
    colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd']  # Distinct colors
    
    for i, color in zip(range(n_classes), colors):
        plt.plot(fpr[i], tpr[i], color=color, lw=2, alpha=0.8,
                 label='ROC curve of {0} (AUC = {1:0.2f})'
                 ''.format(class_names[i], roc_auc[i]))
    
    plt.plot([0, 1], [0, 1], 'k--', lw=2, alpha=0.5)
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate', fontsize=12)
    plt.ylabel('True Positive Rate', fontsize=12)
    plt.title('Receiver Operating Characteristic (ROC) Curves', fontsize=14)
    plt.legend(loc="lower right", fontsize=10)
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig('roc_curves.png', dpi=300)
    plt.show()
    
    return fpr, tpr, roc_auc

# Plot TPR and FPR for each class with smooth curves
def plot_tpr_fpr_curves(fpr, tpr, class_names):
    plt.figure(figsize=(12, 8))
    colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd']  # Distinct colors
    
    for i, color in zip(range(len(class_names)), colors):
        plt.plot(fpr[i], tpr[i], color=color, lw=2, alpha=0.8,
                 label=f'{class_names[i]}')
    
    plt.xlabel('False Positive Rate', fontsize=12)
    plt.ylabel('True Positive Rate', fontsize=12)
    plt.title('True Positive Rate vs False Positive Rate', fontsize=14)
    plt.legend(loc="lower right", fontsize=10)
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig('tpr_fpr_curves.png', dpi=300)
    plt.show()

# Main execution
if __name__ == "__main__":
    # Set random seeds for reproducibility
    random.seed(42)
    np.random.seed(42)
    torch.manual_seed(42)
    
    # Create balanced sample data with 250 samples per class (0-4)
    df = create_balanced_sample_data(SAMPLES_PER_CLASS)
    
    # Preprocess data
    df_processed = preprocess_data(df)
    
    print(f"Total samples: {len(df_processed)}")
    print("Class distribution:")
    print(df_processed['label'].value_counts().sort_index())
    
    # Split data with stratification
    train_df, temp_df = train_test_split(
        df_processed, 
        test_size=0.3, 
        random_state=42, 
        stratify=df_processed['label']
    )
    val_df, test_df = train_test_split(
        temp_df, 
        test_size=0.5, 
        random_state=42, 
        stratify=temp_df['label']
    )
    
    print(f"\nTrain size: {len(train_df)}, Val size: {len(val_df)}, Test size: {len(test_df)}")
    print("Train class distribution:", dict(train_df['label'].value_counts().sort_index()))
    print("Val class distribution:", dict(val_df['label'].value_counts().sort_index()))
    print("Test class distribution:", dict(test_df['label'].value_counts().sort_index()))
    
    # Initialize tokenizer with proper padding token
    tokenizer = GPT2Tokenizer.from_pretrained('gpt2')
    
    # Set padding token - this is crucial for batch processing
    if tokenizer.pad_token is None:
        tokenizer.pad_token = tokenizer.eos_token
        tokenizer.pad_token_id = tokenizer.eos_token_id
    
    print(f"\nTokenizer padding token: {tokenizer.pad_token}")
    print(f"Tokenizer padding token ID: {tokenizer.pad_token_id}")
    
    # Create RAG knowledge base and add training examples
    rag_kb = RAGKnowledgeBase()
    rag_kb.add_examples(train_df['text'].tolist(), train_df['label'].tolist())
    print("RAG knowledge base created with training examples")
    
    # Choose approach: RAG (True) or standard classification (False)
    USE_RAG = False  # Using standard classification for better probability estimates
    
    # Create datasets
    train_dataset = BugReportDataset(
        train_df['text'].tolist(),
        train_df['label'].tolist(),
        tokenizer,
        MAX_LENGTH,
        rag_kb=rag_kb if USE_RAG else None,
        use_rag=USE_RAG
    )
    
    val_dataset = BugReportDataset(
        val_df['text'].tolist(),
        val_df['label'].tolist(),
        tokenizer,
        MAX_LENGTH,
        rag_kb=rag_kb if USE_RAG else None,
        use_rag=USE_RAG
    )
    
    test_dataset = BugReportDataset(
        test_df['text'].tolist(),
        test_df['label'].tolist(),
        tokenizer,
        MAX_LENGTH,
        rag_kb=rag_kb if USE_RAG else None,
        use_rag=USE_RAG
    )
    
    # Create data loaders with num_workers=0 to avoid tokenizer parallelism issues
    train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=0)
    val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)
    test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)
    
    # Initialize model with proper tokenizer
    model = initialize_model(num_labels=5, tokenizer=tokenizer, use_rag=USE_RAG)
    
    # Verify padding token is set correctly
    print(f"Model padding token ID: {model.config.pad_token_id}")
    
    # Optimizer and scheduler
    optimizer = AdamW(model.parameters(), lr=LEARNING_RATE, weight_decay=0.01)
    total_steps = len(train_loader) * EPOCHS
    scheduler = get_linear_schedule_with_warmup(
        optimizer,
        num_warmup_steps=min(WARMUP_STEPS, total_steps),
        num_training_steps=total_steps
    )
    
    # Train model
    print(f"\nStarting {'RAG-based ' if USE_RAG else ''}training...")
    train_losses, val_accuracies = train_model(
        model, train_loader, val_loader, optimizer, scheduler, EPOCHS, use_rag=USE_RAG
    )
    
    # Load best model
    if os.path.exists('best_gpt2_model.pth'):
        model.load_state_dict(torch.load('best_gpt2_model.pth'))
        print("Loaded best model for evaluation.")
    
    # Final evaluation
    test_accuracy = evaluate_model(model, test_loader, use_rag=USE_RAG)
    print(f"\nTest Accuracy: {test_accuracy:.4f}")
    
    # Get predictions and probabilities for detailed analysis
    predictions, probabilities, actual_labels = get_predictions_and_probabilities(
        model, test_loader, use_rag=USE_RAG
    )
    
    # Detailed classification report
    print("\nClassification Report:")
    print(classification_report(actual_labels, predictions, target_names=CLASS_NAMES))
    
    # Plot confusion matrix with percentages
    print("\nGenerating confusion matrix (percentage-wise)...")
    cm = plot_confusion_matrix_percentage(actual_labels, predictions, CLASS_NAMES)
    
    # Plot ROC curves (only for standard classification as RAG doesn't provide proper probabilities)
    if not USE_RAG:
        print("Generating ROC curves...")
        fpr, tpr, roc_auc = plot_roc_curves(actual_labels, probabilities, 5, CLASS_NAMES)
        
        # Plot TPR vs FPR curves
        print("Generating TPR vs FPR curves...")
        plot_tpr_fpr_curves(fpr, tpr, CLASS_NAMES)
    else:
        print("Skipping ROC curves for RAG approach (no proper probability estimates)")
    
    # Plot training history
    plt.figure(figsize=(12, 4))
    
    plt.subplot(1, 2, 1)
    plt.plot(train_losses, label='Training Loss')
    plt.title('Training Loss')
    plt.xlabel('Epochs')
    plt.ylabel('Loss')
    plt.legend()
    
    plt.subplot(1, 2, 2)
    plt.plot(val_accuracies, label='Validation Accuracy', color='orange')
    plt.title('Validation Accuracy')
    plt.xlabel('Epochs')
    plt.ylabel('Accuracy')
    plt.legend()
    
    plt.tight_layout()
    plt.savefig('training_history.png')
    plt.show()
    
    # Demonstrate RAG in action with a few examples
    if USE_RAG:
        print("\n=== RAG Demonstration ===")
        test_examples = test_df.sample(3)
        
        for i, (idx, row) in enumerate(test_examples.iterrows()):
            print(f"\nExample {i+1}:")
            print(f"Bug Report: {row['text']}")
            print(f"True Severity: {CLASS_NAMES[row['label']]}")
            
            # Retrieve similar examples
            retrieved = rag_kb.retrieve_similar(row['text'], top_k=2)
            print("Retrieved similar examples:")
            for j, example in enumerate(retrieved):
                print(f"  {j+1}. [{CLASS_NAMES[example['label']]}] {example['text'][:100]}... (sim: {example['similarity']:.3f})")
            
            # Get prediction
            input_text = tokenizer(
                row['text'],
                truncation=True,
                padding='max_length',
                max_length=MAX_LENGTH,
                return_tensors='pt'
            )
            
            input_ids = input_text['input_ids'].to(device)
            attention_mask = input_text['attention_mask'].to(device)
            
            # Generate with RAG context
            rag_prompt = "Classify the severity of this bug report based on similar examples:\n\n"
            
            for j, example in enumerate(retrieved):
                severity = CLASS_NAMES[example['label']]
                rag_prompt += f"Example {j+1} ({severity}): {example['text']}\n\n"
            
            rag_prompt += f"Bug report to classify: {row['text']}\n\nSeverity:"
            
            rag_input = tokenizer(
                rag_prompt,
                truncation=True,
                padding='max_length',
                max_length=MAX_LENGTH,
                return_tensors='pt'
            )
            
            rag_input_ids = rag_input['input_ids'].to(device)
            rag_attention_mask = rag_input['attention_mask'].to(device)
            
            outputs = model.generate(
                input_ids=rag_input_ids,
                attention_mask=rag_attention_mask,
                max_length=len(rag_input_ids[0]) + 20,
                num_return_sequences=1,
                temperature=0.7,
                do_sample=True,
                pad_token_id=model.config.pad_token_id
            )
            
            generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
            predicted_severity = extract_severity_from_text(generated_text)
            
            print(f"Predicted Severity: {CLASS_NAMES[predicted_severity]}")
            print(f"Generated Text: {generated_text}")
    
    print(f"\nThanks! This {'RAG-based ' if USE_RAG else ''}code is working fine!")
```
